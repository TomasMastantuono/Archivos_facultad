```python
import numpy as np
from scipy.integrate import quad
```

# Estimación de los tiempos de igualdad

```python
O_g = 8.24e-5
O_m = .3
alpha = O_g/O_m
H0 = 2.2E-18 #s-1

def t_eq(z, alpha = alpha, H0 = H0):
    x = 1 + z
    pre_f = 1/(H0 * np.sqrt(O_m))
    fterm = (4/3)*alpha**(3/2)
    sterm = 2 * alpha**(3/2) * np.sqrt( (alpha*x + 1) / (alpha*x) ) * ( 1 - ( (alpha*x + 1)/(3*alpha*x) ) )
    return pre_f*(fterm - sterm)
```

```python
x_mg = 3000
x_mL = 1.33
t_mg = t_eq(x_mg)
t_mL = t_eq(x_mL)

print(t_mg/60/60/24/365)
print(t_mL/60/60/24/365)
```

# Integral numérica

```python
O_L = .7

integrando = lambda x: 1/( x * np.sqrt(O_g*x**4 + O_m*x**3 + O_L) )
val_mg, error = quad(integrando, x_mg, np.inf)
val_mL, error = quad(integrando, x_mL, np.inf)


print(val_mg/H0/60/60/24/365)
print(val_mL/H0/60/60/24/365)
```
