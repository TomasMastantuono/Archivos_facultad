```python
import numpy as np
import pandas as pd
from scipy.integrate import quad

parametros = pd.read_csv('resultados_ajuste.csv', index_col = 0)
H0_ = parametros.query('Cantidad == "H0"')['Valor'].values[0] # km/s/Mpc
mpc_en_metros = 3E19
H0 = H0_/mpc_en_metros
print(H0)

# http://hyperphysics.phy-astr.gsu.edu/hbase/Astro/denpar.html
O_m = .27
O_g = 8.24E-5
O_l = 0.73
O_k = 0
```

```python
integrando = lambda a: 1/(a*np.sqrt(O_m/(a**3) + O_g/(a**4) + O_l))
val_int, err = quad(integrando, 0, 1)
print(val_int/H0/60/60/24/365/1E9)
print(err)
```
